# Loading the Data

About the data: https://www.gapminder.org/tools/#$chart-type=bubbles&url=v2

```python
import plotly.offline as py
py.init_notebook_mode(connected=True)
import plotly.graph_objs as go
import pandas as pd
import numpy as np
```

```python
import plotly.express as px
from plotly.figure_factory import create_table
```

```python
# Importing built-in dataset
gapminder = px.data.gapminder()
table = create_table(gapminder.head(10))
py.iplot(table)
```

# Quick Visualizations with Custom Bar Charts

```python
data_brazil = px.data.gapminder().query("country == 'Brazil'")
fig = px.bar(data_brazil, x='year', y='pop', height=400)
fig.show()
```

```python
# Plot population of Brazil over the years, with life expectancy as color and hover data
fig = px.bar(data_brazil, x='year', y='pop',
            hover_data=['lifeExp', 'gdpPercap'], color='lifeExp',
            labels={'pop': "Population of Brazil",'year': "Year", 'lifeExp': "Life Expectancy"}, height=400)
fig.show()
```

# Plot Life Expectancy vs GDP per Capita

```python
gapminder2007 = gapminder.query("year == 2007")
px.scatter(gapminder2007, x='gdpPercap', y='lifeExp')
```

```python
px.scatter(gapminder2007, x='gdpPercap', y='lifeExp', color='continent')
```

# Customize Interactive Bubble Charts

```python
px.scatter(gapminder2007, x='gdpPercap', y='lifeExp', color='continent',
          size='pop', size_max=60)
```

```python
px.scatter(gapminder2007, x='gdpPercap', y='lifeExp', color='continent',
          size='pop', size_max=60, hover_name='country')
```

# Create Interactive Animations and Facet Plots

```python
px.scatter(gapminder2007, x='gdpPercap', y='lifeExp', color='continent',
          size='pop', size_max=60, hover_name='country', facet_col='continent', log_x=True)
```

```python
px.scatter(gapminder, x='gdpPercap', y='lifeExp', color='continent',
          size='pop', size_max=60, hover_name='country',
          animation_frame='year',
          animation_group='country',
          log_x=True,
          range_x=[100,100000],
          range_y=[25,90],
          labels=dict(pop="Population", gdpPercap="GDP per Capita", lifeExp="Life Expectancy")
)
```

# Represent Geographic Data as Animated Maps

```python
px.choropleth(gapminder,
              locations="iso_alpha",
              color="lifeExp",
              hover_name="country",
              animation_frame="year",
              color_continuous_scale=px.colors.sequential.Plasma,
              projection="natural earth")
# projection="orthographic"
```
